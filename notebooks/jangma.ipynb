```python
import matplotlib.pyplot as plt     # Plotting package
import numpy as np                      # Scientific computing package
import xarray as xr                     # NetCDF data package
import pandas as pd                     # table type data analysis package
```

```python
#-------- Reading data ----------
data = pd.read_csv('../data/Jangma_Jeonju_1980-2025.csv', encoding='euc-kr')
data.head()
```

```python
y1 = data['장마일수']
y2 = data['강수일수']
y3 = data['합계강수량']
```

```python
# 방법 1: plt.plot() — 간단하지만 그래프가 여러 개면 곧 한계
# 방법 2: 도화지(figure)와 그림 칸(axes)을 먼저 만들고, 칸 위에 그리기
fig = plt.figure() # 도화지
ax = plt.axes()  # ax  : 도화지 위의 그래프 칸 하나 
# fig, ax = plt.subplots()   # 위 두 줄을 한 번에 
```

```python
fig, axes = plt.subplots(2, 3, figsize=(8, 4.5))   # 2행 3열 → 칸 6개, axes는 2×3 배열
fig.suptitle("fig, axes = plt.subplots(2, 3)  ->  axes[row, col]")

for i in range(2):          # 행 번호 (위→아래, 0부터)
    for j in range(3):      # 열 번호 (왼쪽→오른쪽, 0부터)
        ax = axes[i, j]     # 각 칸 꺼내기
        ax.text(0.5, 0.5, f"axes[{i}, {j}]", ha='center', va='center', fontsize=14)
        ax.set_xticks([]); ax.set_yticks([])   # 눈금 지워서 번호만 보이게

plt.tight_layout()   # 칸끼리 안 겹치게 간격 자동 조정
plt.show()
```

```python
#------- Draw Line Graph
# plt.plot(...)  →  ax.plot(...)   앞에 붙는 이름만 ax로 바뀜, 옵션은 그대로
fig, ax = plt.subplots() 
ax.plot(x, y2, color='r', marker='o', linestyle='dashed', linewidth=1.5, label='Precip Days')
```

```python
fig, ax = plt.subplots() 
ax.bar(x, y2, color='g', width=0.72, label='Precip Days')
```

```python
#--- Setting title, label, legend
# 꾸미기 명령은 이름이 조금 달라짐 (set_ 이 붙음)
#   plt.title()  → ax.set_title()
#   plt.xlabel() → ax.set_xlabel()
#   plt.legend() → ax.legend()      (legend는 set_ 없음)
ax.set_title('Jeonju_Jangma Precip Days')
ax.set_xlabel('Time')
ax.legend(loc='upper left')
plt.show()   # 출력은 여전히 plt.show(), 마지막에 한 번
```

```python

#---------- Plotting ------------
# figsize=(가로, 세로) 인치 / dpi: 해상도 (클수록 선명, 파일도 커짐)
fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
ax2 = ax.twinx()   
#--- Graph 1 : 장마일수 (왼쪽 y축, 막대)
ax.bar(x, y1, color='gray', width=0.72, label='Jangma Days')
ax.set_ylabel('Days')                 
ax.set_title('Jeonju Jangma data')


ax.plot(x, y2, color='g', marker='o', linestyle='dashed', linewidth=1.5, label='Precip Days')
ax.legend(loc='upper left')           

ax2.plot(x, y3, color='b', marker='o', linestyle='dashed', linewidth=1.5, label='Precipitation')
ax2.set_ylabel('Precipitation (mm)', color='b')  

ax2.spines['right'].set_color('b')                           
ax2.tick_params(axis='y', color='b', labelcolor='b')        
# ax2.spines['right'].set_position(('axes', 1.0))  # 축 위치 (1.0은 기본값이라 생략 가능,
                                                   
ax2.legend(loc='lower right')         
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(13, 4), dpi=150)

#--- 칸 0 : 장마일수 (막대)
axes[0].bar(x, y1, color='gray', width=0.72, label='Jangma Days')
axes[0].set_title('Jangma Days')
axes[0].set_ylabel('Days')

#--- 칸 1 : 강수일수 (선)
axes[1].plot(x, y2, color='g', marker='o', linestyle='dashed', linewidth=1.5, label='Precip Days')
axes[1].set_title('Precip Days')
axes[1].set_ylabel('Days')

#--- 칸 2 : 합계강수량 (선) — 단위가 mm라 따로 그려야 함
axes[2].plot(x, y3, color='b', marker='o', linestyle='dashed', linewidth=1.5, label='Precipitation')
axes[2].set_title('Precipitation')
axes[2].set_ylabel('mm')

#--- 세 칸 공통 설정은 반복문으로
for ax in axes:
    ax.set_xlabel('Year')
    ax.legend(loc='upper left')

fig.suptitle('Jeonju Jangma (1980-2025)')   
plt.tight_layout()
plt.show()
```
